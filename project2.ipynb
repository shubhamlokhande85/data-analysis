### 21. Matplotlib – Bar Chart

```python
x=df["Name"]
y=df["Final_Score"] 
plt.figure(figsize=(15,5))
plt.bar(x,y)
plt.title("Student Final Score")
plt.ylabel("Final_Score")
plt.xlabel("Students")
plt.xticks(rotation=50)
plt.show()
```

### 22. Matplotlib – Line Plot


```python
x=df["Study_Hours"]
y=df["Final_Score"]
plt.figure(figsize=(15,5))
plt.plot(x,y,marker="o")
plt.title("Study Hours VS Final Score")
plt.xlabel("Study Hours")
plt.ylabel("Final Score")
plt.grid()
plt.show()
```

### 23. Matplotlib – Scatter Plot

```python
x=df["Study_Hours"]
y=df["Final_Score"]
plt.figure(figsize=(15,5))
plt.scatter(x,y)
plt.title("Study Hours VS Final Score")
plt.xlabel("Study Hours")
plt.ylabel("final score")
plt.show()
```

### 24. Matplotlib – Histogram

```python
plt.figure(figsize=(15,5))
x=df["Final_Score"]
plt.hist(x, bins=5)
plt.title("Distribution of final score ")
plt.ylabel("No of students")
plt.xlabel("Final Score")
plt.show()
```

### 25. Matplotlib - Pie Chart 

```python
count_gender = df["Gender"].value_counts()
plt.pie(count_gender, labels=count_gender.index, autopct="%1.1f%%")
plt.title("gender distribution")
plt.show()
```

### 26. Seaborn – Count Plot

```python
sns.countplot(df,x="Gender")
plt.title("No of Male and Female students")
plt.show()
```

### 27. Seaborn Barplot 

```python
sns.barplot(df,x="Gender",y="Final_Score")
plt.title("Average final score by gender ")
plt.show()
```

### 28. Seaborn – Scatter Plot with Hue

```python
sns.scatterplot(df,x="Study_Hours",y="Final_Score",hue="Gender")
plt.title("Gender wise study hours and final score")
plt.show()
```

### 29. Seaborn – Box Plot


```python
sns.boxplot(data=df[["Maths","Science","English"]])
plt.title("Subject wise Mark distribution")
plt.show()
```

### 30.Seaborn - heatmap

```python
correlation = df.select_dtypes("number").corr()
sns.heatmap(correlation,annot=True)
plt.title("correlation heatmap")
plt.show()
```

### 31. Pairplot

```python
sns.pairplot(df[["Study_Hours","Attendance","Final_Score"]])
plt.show()
```

### 32. Saving matplotlib plot 

```python
plt.figure(figsize=(15,5))
plt.bar(df["Name"],df["Final_Score"])
plt.title("students final score")
plt.xlabel(" StudentNames")
plt.xticks(rotation=90)
plt.ylabel("Score")
plt.savefig("STD_SCORE.png")
plt.show()
```
