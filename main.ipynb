### 23.  Matplotlib Bar Chart  - Sales by Categories 

```python
plt.figure(figsize=(10,5))
category_sales=df.groupby("Category")["Total_Sales"].sum()
plt.bar(category_sales.index, category_sales.values)
plt.title("Sales By Category ")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

### 24.Matplotlib Line Chart  - Sales Trends Over a Period 

```python
plt.figure(figsize=(20,5))
x = df["Date"]
y = df["Total_Sales"]
plt.plot(x,y,marker="*nb")
plt.title("Sales Trend")
plt.xlabel("Date")
plt.xticks(rotation=90)
plt.ylabel("Sales")
plt.show()
```

### 25. Matplotlib Histogram - Sales Distribution 


```python
plt.figure(figsize=(15,7))
plt.hist(df["Total_Sales"],bins=5)
plt.title("Sales Distribution ")
plt.xlabel("Sales")
plt.ylabel("Orders")
plt.show()
```

### 26. Matplotlib Scatter Plot - Quantity VS Sales 

```python
plt.figure(figsize=(15,7))
x = df["Quantity"]
y=df["Total_Sales"]
plt.scatter(x,y)
plt.title(" Quantity vs Sales")
plt.xlabel("Quantity")
plt.ylabel("Sales")
plt.show()
```

### 27. Matplotlib Pie Chart - Payment Method Distribution 

```python
plt.figure(figsize=(15,7))
payment = df.groupby("Payment_Method")["Total_Sales"].sum()
# print(payment)
plt.pie(payment,labels=payment.index,autopct="%1.1f%%")
plt.title("Sales by Payment Method")
plt.show()
```

### 28. Seaborn Count Plot - Order by Category 

```python
plt.figure(figsize=(15,7))
sns.countplot(data=df, x="Category")
plt.title("Order by Category ")
plt.show()
```

### 29. Seaborn Bar Plot - Sales By City 

```python
plt.figure(figsize=(15,7))
sns.barplot(data=df,x="City",y="Total_Sales")
plt.title("Total sales by city")
plt.show()
```

### 30.Seaborn Box Plot - Sales Distribution by Category 

```python
plt.figure(figsize=(15,7))
sns.boxplot(data=df, x = "Category", y="Total_Sales")
plt.title("Sales Distribution by Category")
plt.show()
```

###  31. Seaborn Scatter Plot - Quantity Vs Sales 

```python
df.head()
plt.figure(figsize=(15,7))
sns.scatterplot(data=df, x = "Quantity", y= "Total_Sales")
plt.title("Quantity Vs Sales ")
plt.show()
```

### 32. Seaborn Pair plot - Correletion of Data 

```python
df.head()
plt.figure(figsize=(15,5))
sns.pairplot(df[["Quantity","Unit_Price", "Total_Sales"]])
plt.show()

```

### 32. Seaborn Heatmap  - Correletion Of Data

```python

plt.figure(figsize=(15,7))
corr=df.select_dtypes("number").corr()
sns.heatmap(corr, annot=True )
plt.show()
```
