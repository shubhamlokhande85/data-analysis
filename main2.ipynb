## Import required libraries 

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
import os 
```

## Import Dataset 

```python
df=pd.read_csv("supermarket_sales.csv")
```

### 1. Which city has the highest total sales?

```python
city=df.groupby("City")["Total_Sales"].sum()
print(city)
print("The highest sales had Pune  ",city.max())
```

### 2. Which product generated the highest sales?

```python
product=df.groupby("Product")["Total_Sales"].sum()
print(product)
print(product.max())
print("The Laptop generated highest sale over the period, sales = " ,product.max())
```

### 3. Which category has the most orders?

```python
df.head()
category=df.groupby("Category")["Category"].value_counts()
print(category)
print(f"The Supermarket Electronic Category Have Most Orders , Electronic = ", category.max())
```

### 4. Which payment method is most common?

```python
df.head()
pay_method=df.groupby("Payment_Method")["Order_ID"].sum()
print(pay_method)
print("The Online payment are Mostly common method in supermarket, Online Transactions =  ", pay_method.max())

```

### 5. Does quantity have a relationship with total sales?

```python
corr =df["Quantity"].corr(df["Total_Sales"])
print(corr)
print("The Sales is Lightly Dependent on Quantity, Correlation between Quantity and Total Sales = ",corr)
```

### 6. Which category has the highest average order value

```python
avg_od = df.groupby("Category")["Total_Sales"].mean()
print(avg_od)
print("The Electronics category have most average order values ,", avg_od.idxmax(),avg_od.max())
```

### 7.Which are the top 5 orders by sales?

```python
df.nlargest(5,"Total_Sales")
```

### 8. What does the correlation heatmap tell you

```python
corr =df.select_dtypes("number").corr()
plt.figure(figsize=(15,6))
sns.heatmap(corr, annot=True)
```
