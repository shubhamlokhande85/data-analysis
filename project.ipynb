# Numpy & Pandas Operations with Code Snippets 


## 1. Import all required libraries 

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
import os 
```

## 2. Importing dataset 

```python
df = pd.read_csv("student_numpy_pandas_practice.csv")
```

##  3. Understanding Dataset 

```python
print(df.shape)
print(df.head())
print(df.tail())
print(df.columns)
print(df.dtypes)
print(df.describe)
```

## 4. Selecting a single column from dataset


```python
names = df["name"]
print(names)
```

## 5. Selecting multiple columns from dataset 

```python
print(df[["name", "age","city"]])
```

## 6. Indexing with iloc - Selecting a rows and columns from dataset 


```python
print(df.iloc[0])
```

```python
print(df.iloc[0:5])
```

```python
print(df.iloc[2:5 , 0:5])
```

## 7. Indexing with loc  -  selecting row and column using row_index_number & column name 

```python
print(df.loc[0, "name"])
```

```python
print(df.loc[0:2, ["name","science"]])
```

## 8. Filtering with conditions

```python
# return student whose science marks are greater than 80 
high_sci = df[df["science"]> 80]
print(high_sci)
```

## 9.1  Multiple conditions - Using AND logical condition 

```python
# return student mark are above 80 and also their attendence above 90 
result2 = df[
    (df["science"]> 80 )
    & 
    (df["attendance"]> 90 ) 
    ]
print(result2)
```

## 9.2 Multiple condition - Using OR logical condition 

```python
# return student details where their marks are  less than 80 in math or less than 80 in science 

result3= df [
    (df["math"]< 80 ) 
    | 
    (df["science"] < 80 )
]

print(result3)
```

## 10.  Numpy array from a column - converitng pandas column into numpys array 

```python
science_marks = df["science"].to_numpy()
print(science_marks)
print(type(science_marks))
```

## 11. NumPy Minimum and Maximum - reteriving minimum and maximum value form array 

```python

print(np.min(science_marks))
print(np.max(science_marks))
```

## 12. NumPy argmin and argmax - reteriving min value position and max value position from array 


```python
print(np.argmin(science_marks))
print(np.argmax(science_marks))
print(df.iloc[np.argmax(science_marks)])
```

## 13 . Mean,Meadian & Standard Deviation  - Applying numerical statistics on array 

```python
print(np.mean(science_marks))
print(np.median(science_marks))
print(np.std (science_marks))

```

## 14. Vectorized operations 
 

```python
print(science_marks + 5)
print(science_marks * 5)
print(science_marks / 5)
```

## 15 .Boolean mask in numpy 

```python
mask = science_marks > 80 
print(mask)
print(science_marks[mask])
```

## 16 . Create Total & Average Columns 

```python
df["total"] = df["math"] + df["science"]
df["average"] = df["total"]/2
print(df[["name","total","average"]])
```

## 17. Create Result Column 

```python
df["results"] = np.where(df["average"]>=35 , "pass", "fail")
print(df[["name", "average","results"]])
```

## 18.Sort data

```python
# sort row based on column 
# displaying top 5 students from science 

sorted_df = df.sort_values("science" , ascending=False )
print(sorted_df.head())
```

## 19. Top 5 students using nlargest() 


```python
top5 = df.nlargest(5,"average")
print(top5[["name","math","science","average"]])
```

## 20 . GroupBy - Average by city 


```python
average_city = df.groupby("city")["science"].mean()
print(average_city)
```

## 21. GroupBy - Multiple Statistics 

```python
summary = df.groupby("city")["science"].agg(["mean","max", "max","median","std","count"])
print(summary)
```

## 22. Missing values

```python
print(df.isnull().sum())
print(df[df.isnull().any(axis=1)])
```

## 23 . Fill Missing values

NOTE - No Missing values in dataset but still for exploration of handling missing values 

```python
df["science"] = df["science"].fillna(df["science"].mean())
```

## 24. Selecting Students by city - Using categorical column


```python
mumbai_students = df[df["city"]=="Mumbai"]
print(mumbai_students)
```

## 25 . find highest attendance student - using idxmax()



```python
index = df["attendance"].idxmax()
print(index)
print(df.loc[index,["name","attendance"]])
```

## 26 . find lowest attendance student - using idxmin()


```python
index =df["attendance"].idxmin()
print(index)
print(df.loc[index,["name","attendance"]])
```

## 27. Correlation 

```python
print(df[["math","science","attendance","study_hours"]].corr())
```
